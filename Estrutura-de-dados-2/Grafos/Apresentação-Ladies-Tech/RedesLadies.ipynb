```python
rede = {
    "fernandagil": ["lauraaz", "Saara.lima", "maria_anq"],
    "Saara.lima": ["fernandagil", "emily.dev", "isa.linhares"],
    "emily.dev": ["Saara.lima", "leticiagly"],
    "leticiagly": ["emily.dev", "isa.linhares", "ths.llys"],
    "ths.llys": ["leticiagly", "annanadroz", "ysadox"],
    "ysadox": ["eusolz", "ths.llys"],
    "eusolz": ["thalyana", "annanadroz", "ysadox"],
    "thalyana": ["lauraaz", "maria_anq", "eusolz"],
    "lauraaz": ["fernandagil", "thalyana"],
    "maria_anq": ["fernandagil", "thalyana", "annanadroz"],
    "annanadroz": ["maria_anq", "isa.linhares", "eusolz", "ths.llys"],
    "isa.linhares": ["Saara.lima", "leticiagly", "annanadroz"]
}
n = len(rede)
soma_graus = sum(len(a) for a in rede.values())
E = soma_graus // 2
D = 2 * E/(n * (n - 1))
print ("A quatidade de usuario é:",n ,"De redes é:",E, "E a densidade é de:",f"{D:.2%}")
print(n, E, f"{D:.2%}")
```

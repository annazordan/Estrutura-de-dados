# Problema real — Sistema de atendimento de uma clínica

```python
class Paciente:
    def __init__(self, nome, idade, prioridade):
        self.nome = nome
        self.idade = idade
        self.prioridade = prioridade

class Node:
    def __init__(self, paciente):
        self.paciente = paciente
        self.proximo = None

class FilaAtendimento:
    def __init__(self):
        self.primeiro = None
        self.ultimo = None
        self.tamanho = 0

    # Adicionar paciente (no final da fila)
    def adicionar(self, paciente):
        novo_no = Node(paciente)
        if self.esta_vazia():
            self.primeiro = novo_no
            self.ultimo = novo_no
        else:
            self.ultimo.proximo = novo_no
            self.ultimo = novo_no
        self.tamanho += 1

    # Listar pacientes
    def listar(self):
        if self.esta_vazia():
            print("A fila está vazia.")
            return

        atual = self.primeiro
        print("\n--- Lista de Pacientes na Fila ---")
        while atual is not None:
            p = atual.paciente
            print(f"{p.nome} — {p.idade} anos — {p.prioridade}")
            atual = atual.proximo
        print("----------------------------------\n")

    def atender(self):
        if self.esta_vazia():
            print("Não há pacientes para atender.")
            return None

        paciente_atendido = self.primeiro.paciente
        self.primeiro = self.primeiro.proximo

        if self.primeiro is None:
            self.ultimo = None

        self.tamanho -= 1
        return paciente_atendido

    # Verificar se a fila está vazia
    def esta_vazia(self):
        return self.primeiro is None

    def quantidade(self):
        return self.tamanho



fila = FilaAtendimento()

fila.adicionar(Paciente("Ana", 32, "Normal"))
fila.adicionar(Paciente("Bruno", 70, "Prioridade"))
fila.adicionar(Paciente("Carlos", 45, "Normal"))

print(f"Pacientes na fila no momento: {fila.quantidade()}")
fila.listar()

paciente = fila.atender()
if paciente is not None:
    print(f"Atendendo: {paciente.nome}")

print(f"Pacientes restantes na fila: {fila.quantidade()}")
fila.listar()

if fila.esta_vazia():
    print("A fila agora está vazia.")
else:
    print("Ainda há pacientes aguardando.")
```

### O que é uma referência em Python?

### Qual a diferença entre b = a e b = a.copy()?

no outro, "b" copia "a", assim, pode adcionar mais elementos.

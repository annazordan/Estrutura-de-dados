# Exercicio 01
## O simulador "Desfazer"

```python
class EditorDeTexto:
    def __init__(self):
        self.historico = []

    def executar_acao(self, tipo, conteudo=None):
        if tipo == 'digitar':
            # Guarda o que foi digitado
            self.historico.append(conteudo)

        elif tipo == 'apagar':
            # Apaga a última coisa
            if len(self.historico) > 0:
                self.historico.pop()

        elif tipo == 'substituir':
            # Substitui a última ação
            if len(self.historico) > 0:
                self.historico[-1] = conteudo

        elif tipo == 'desfazer':
            # Remove e desfaz
            if len(self.historico) > 0:
                self.historico.pop()

        else:
            print("Ação inválida!")

    def mostrar_historico(self):
        print("Histórico:", self.historico)


# Usando
editor = EditorDeTexto()

editor.executar_acao('digitar', 'Ola')
editor.executar_acao('digitar', 'Mundo')

editor.mostrar_historico()

editor.executar_acao('desfazer', None)

editor.mostrar_historico()
```

# Exercicio 02
## Sistema de impressao

```python
class SistemaDeImpressao:
    def __init__(self):
        self.fila = []

    def adicionar_documento(self, documento):
        # Adiciona o documento no final
        self.fila.append(documento)

    def imprimir_documento(self):
        # Imprime o documento mais antigo
        if len(self.fila) > 0:
            documento = self.fila.pop(0)
            print("Imprimindo:", documento)
        else:
            print("Não há documentos na fila.")

    def mostrar_fila(self):
        print("Fila de impressão:", self.fila)


# Uando
impressora = SistemaDeImpressao()

impressora.adicionar_documento("Trabalho ")
impressora.adicionar_documento("Relatório")
impressora.adicionar_documento("Exercício")

impressora.mostrar_fila()

impressora.imprimir_documento()
impressora.imprimir_documento()

impressora.mostrar_fila()
```
